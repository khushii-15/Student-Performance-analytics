````python
def performance_category(marks):
    if marks >= 85:
        return "High"
    elif marks >= 70:
        return "Medium"
    else:
        return "Low"

df["Performance_Category"] = df["Final_Marks"].apply(performance_category)

print(df)
````

````python
performance_counts = df["Performance_Category"].value_counts()

print(performance_counts)
````

````python
plt.figure(figsize=(7, 5))

performance_counts.plot(kind="bar")

plt.xlabel("Performance Category")
plt.ylabel("Number of Students")
plt.title("Student Performance Distribution")

plt.show()
````

````python
top_3_students = df.sort_values(
    by="Final_Marks",
    ascending=False
).head(3)

print(top_3_students)
````

````python
top_3_students = df.sort_values(
    by="Final_Marks",
    ascending=False
).head(3)

print(top_3_students)
````

````python
category_average = df.groupby("Performance_Category")["Final_Marks"].mean()

print(category_average)
````

````python
plt.figure(figsize=(7, 5))

category_average.plot(kind="bar")

plt.xlabel("Performance Category")
plt.ylabel("Average Final Marks")
plt.title("Average Final Marks by Performance Category")

plt.show()
````

````python
def check_student_performance(student_id):
    student = df[df["Student_ID"] == student_id]

    if student.empty:
        print("Student ID not found.")
        return

    print(student.to_string(index=False))
````

````python
check_student_performance(108)
````

````python
def give_recommendation(student_id):
    student = df[df["Student_ID"] == student_id]

    if student.empty:
        print("Student ID not found.")
        return

    marks = student["Final_Marks"].iloc[0]
    attendance = student["Attendance"].iloc[0]
    study_hours = student["Study_Hours"].iloc[0]

    print("Student ID:", student_id)
    print("Final Marks:", marks)
    print("Attendance:", attendance, "%")
    print("Study Hours:", study_hours)

    if marks >= 85:
        print("Recommendation: Excellent performance. Keep maintaining your study routine.")
    elif marks >= 70:
        print("Recommendation: Good performance. Focus on improving weak areas.")
    else:
        print("Recommendation: Needs improvement. Increase study time and focus on assignments.")
````

````python
give_recommendation(103)
````
